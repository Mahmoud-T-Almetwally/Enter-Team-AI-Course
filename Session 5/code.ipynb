```python
import pandas as pd

# we pass 'index_col' to avoid redundancy since "PassengerId" is an index column same as the one pandas adds.
titanic = pd.read_csv("titanic.csv", index_col="PassengerId")

titanic.head() # Inspect the data
```

```python
titanic.info() # get more details of about the dataset
```

```python
# first we will handle the 'Age' column
titanic.Age.describe()
```

```python
# We can the min value being a fraction, this probably means this person was a newborn.
# Let's confirm this by seeing if they had any relatives
titanic[titanic.Age < 1] # Our assumption was correct they all have a relative on board with them.
```

```python
# What about the 'NaN' ones?
titanic[titanic.Age.isna()] # they seem to be randomly missing values.
```

```python
# in this case we could remove outlier values and calculate a mean to fill in the values.
# We will use the IQR method we used in the previous session.
q1 = titanic.Age.quantile(0.25)
q3 = titanic.Age.quantile(0.75)

iqr = q3 - q1

lower_bound = q1 - 1.5 * iqr
upper_bound = q3 + 1.5 * iqr

print(f"Lower Bound: {lower_bound}")
print(f"Upper Bound: {upper_bound}")

# get the values that are greater than the lower bound and lower than the upper bound
in_range_filter = (titanic.Age > lower_bound) & (titanic.Age < upper_bound)
titanic[in_range_filter]
```

```python
# Calculate the mean
age_mean = titanic[in_range_filter].Age.mean()

# fill the missing values, use this method in this way to avoid a deprecation warning.
titanic.fillna({"Age" : age_mean}, inplace=True)
```

```python
# Let's the check Age column now
titanic.Age.describe()
```

```python
# next we will handle the 'Cabin' column
titanic.Cabin.describe() # this column represents the cabin (part of the ship) the passenger was on.
```

```python
titanic.Cabin.value_counts() # let's check the most common values
```

```python
# Let's see how many unique values there are
len(titanic.Cabin.unique()) - 1 # take out the 'NaN' value
```

```python
# The 'Cabin' column seems to be an annoying column to deal with
# it has too many unique values to one-hot encode it.
# and they are evenly distrbuted so filling the missing values with the most common wouldn't be correct
# thus we will just drop it.

titanic.drop(columns=['Cabin'], inplace=True)

titanic.info()
```

```python
# lastly the 'Embarked' column is missing exactly 2 values
titanic[titanic.Embarked.isna()]
```

```python
# Let's check the unique values in the column
titanic.Embarked.unique()
```

```python
# the most common values
titanic.Embarked.value_counts()
```

```python
# we can simply drop these two as well or fill in with the most common value as well.
# both are viable options so we will leave them here:

# in this case we will not keep the 2 rows
# titanic.fillna({"Embarked":titanic.Embarked.mode()}, inplace=True)

# Use "subset" argument to determine on which columns to look for missing values
titanic.dropna(subset=['Embarked'], inplace=True)
```

```python
# Next we'll encode the column

embarked = titanic.pop("Embarked")

embarked_one_hot_encoded = pd.get_dummies(embarked)

embarked_one_hot_encoded.head()
```

```python
# then, simply join them back

cleaned_titanic = pd.concat([embarked_one_hot_encoded, titanic], axis=1)

cleaned_titanic.head()
```

```python
# since the 'Name' column doesn't realy tell the model anything useful and has all unique values we will drop it as well
cleaned_titanic.drop(columns=["Name"], inplace=True)

cleaned_titanic.info()
```

```python
# the 'Sex' column has 2 possible values, so it can be easliy one-hot encoded

# first we pop it
sex_column = cleaned_titanic.pop("Sex")

# then encode it
sex_one_hot_encoded = pd.get_dummies(sex_column)

sex_one_hot_encoded.head()
```

```python
# Then we add back the column in encoded form

final_titanic = pd.concat([sex_one_hot_encoded, cleaned_titanic], axis=1)

final_titanic.head()
```

```python
# Next, the ticket column
final_titanic.Ticket.describe()
```

```python
# since it also doesn't entail any useful info, and is similarly hard to deal with like the cabin column
# we'll drop it

final_titanic.drop(columns=['Ticket'], inplace=True)

final_titanic.info()
```

```python
final_titanic.head()
```

```python
y = final_titanic.pop("Survived") # this is the value we want to predict.

X = final_titanic # all other columns are input to out model.

X.shape, y.shape
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2) # We will use the same test size as last time.

print(f"Train dataset shapes: {X_train.shape, y_train.shape}") 
print(f"Test dataset shapes: {X_test.shape, y_test.shape}") 
```

```python
# Then we use the same model we instantiated 

model.fit(X_train, y_train) # ignore the warning for now
```

```python
# predict on the test data

y_pred = model.predict(X_test)
```

```python
from sklearn.metrics import accuracy_score

# we use it the same way we did with "MSE" (mean squared error) and "MAE" (mean absolute error)
accuracy = accuracy_score(y_test, y_pred)
print(f"We got: {accuracy*100:.2f}% right!")
```
