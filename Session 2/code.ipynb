```python
import numpy as np # importing NumPy as "np" is a standard and reduces code length by a bit
```

```python
normal_list = [1, 2, 3, 4, 5]
nd_array = np.array(normal_list)
print(f"Normal list: {normal_list}, n D Array: {nd_array}")
print(f"Normal list: {type(normal_list)}, n D Array: {type(nd_array)}")
```

```python
nd_array.mean() # Calculates the average
```

```python
nd_array.max(), nd_array.min() # Gets the minimum and maximum values
```

```python
nd_array.sum() # Calculates the summation across the whole array
```

```python
nd_array + 1 # Adds one to each element in the array
nd_array - 1 # Subtracts one from each element in the array
nd_array * 1 # Multiplies each element from the array by 1
nd_array / 1 # Divides each element from the array by 1
# The output will remain the same since we did nothing 
```

```python
nd_array.shape # Returns the dimensions of the array, since there is only one dimension and it has five elements, the shape is (5,)
```

```python
nd_array.dtype # Returns the type of the data stored in the array
```

```python
nd_array.ndim # Returns the number of dimensions
```

```python
arr1 = np.random.randint(low=1, high=10, size=(2, 3))
arr1
```

```python
arr2 = np.random.randint(low=1, high=10, size=(3, 4))
arr2
```

```python
arr1 @ arr2 # matrix of size 2 x 3 with 3 x 4 => a matrix of size 2 x 4 
```

```python
arr1.dot(arr2) # same result
```

```python
(arr1 @ arr2) == (arr1.dot(arr2)) # sanity check
```
