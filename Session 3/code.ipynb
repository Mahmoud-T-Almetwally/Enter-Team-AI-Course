```python
mask = titanic['Survived'] == 1 # first we select the column we want to compare values on, then we store it in a mask
mask.head() # a mask is a 'pd.Series' object with True and False values
```

```python
# titanic[titanic['Survived'] == 1] this is also valid, but introduces complications in the next section
titanic[mask] # then we pass the mask as an index to the dataframe
# what pandas does behind the scenes is keeping values where the index in the mask is True and remove them if False 
```

```python
survived_mask = titanic['Survived'] == 1
male_mask = titanic['Sex'] == "male"

titanic[survived_mask & male_mask]  # this yeilds rows only where the person is a male and survived
```

```python
titanic['Age'].isna() # this returns a mask with rows that have a missing age value
```

```python
titanic[titanic['Age'].isna()] # this returns rows with missing age values
```

```python
titanic['Age'].dropna() # this removed 178 rows!
```

```python
# titanic['Age'].dropna(inplace=True) # we won't be removing values yet
```

```python
average_age = titanic['Age'].mean() # this calculates the average age
titanic['Age'].fillna(average_age) # this returns a 'pd.Series' object with the 'NaN' values filled
```

```python
# Example on a string column:
# the 'dropna' argument tells pandas to calculate the mode without considering 'NaN' values
most_common_peer = titanic['Embarked'].mode(dropna=True) # This returns the most common peer people "Embarked" from.
titanic['Embarked'].fillna(most_common_peer) # this returns a 'pd.Series' object with the 'NaN' values filled
```

```python
plt.hist(x=survived['Sex']) # equivelant code in matplotlib
plt.show() 
```

```python
survived['Sex'].value_counts().plot.pie() # we use value counts to indicate the comparison between the 'number' of male to females
plt.show() # we use 'plt.show()' to seperate the plots and avoid returning a plot object and printing it
```

```python
plt.pie(x=survived['Sex'].value_counts(), labels=survived['Sex'].unique()) # equivelant code in matplotlib
plt.show()
```

```python
survived['Age'].value_counts().sort_index().plot(kind="bar", xticks=[0, 10, 20, 30, 40, 50, 70, 80])
plt.show()
```

```python
plt.bar(x=survived['Age'].value_counts().index, height=survived['Age'].value_counts().values)
plt.show()
```
